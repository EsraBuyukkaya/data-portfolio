## Data Notes

The World Cup winner and runner-up data covers men's FIFA World Cups from 1930 through 2022. The player earnings dataset uses Forbes 2025 highest-paid soccer player estimates.

For the phrase **most popular teams**, this notebook uses a visibility proxy instead of pretending to measure global fan opinion directly:

`visibility_score = titles * 5 + finals * 3 + appearances`

This is more honest because popularity can mean many things, including fans, TV attention, shirts sold, social media, tournament history, and star players.

```python
import pandas as pd

editions = pd.read_csv('data/world_cup_editions.csv')
teams = pd.read_csv('data/team_summary.csv')
players = pd.read_csv('data/highest_paid_players_2025.csv')

editions.head()
```

## Data Wrangling

Before analyzing, I checked the datasets for missing values, duplicate rows, and consistent country names. Germany and West Germany are combined as Germany for this portfolio analysis so the country-level history is easier to compare.

```python
print('World Cup editions:', editions.shape)
print('Team summary:', teams.shape)
print('Highest-paid players:', players.shape)

editions.isna().sum(), teams.isna().sum(), players.isna().sum()
```

## Question 1: Who Won the Most World Cups?

```python
title_leaders = teams.sort_values('titles', ascending=False)[['team', 'region', 'titles', 'finals', 'appearances']]
title_leaders.head(10)
```

**Finding:** Brazil is the all-time leader with five titles. Germany and Italy follow with four each, while Argentina has three after its 2022 win.

![World Cup title leaders](assets/charts/world-cup-title-leaders.svg)

## Question 2: Who Reaches Finals Most Often?

```python
finals_leaders = teams.sort_values('finals', ascending=False)[['team', 'titles', 'finals', 'appearances']]
finals_leaders.head(10)
```

**Finding:** Germany has the strongest finals consistency in this dataset. Brazil has more titles, but Germany has reached the final more often.

## Question 3: Which Teams Are Most Visible?

```python
teams['visibility_score'] = teams['titles'] * 5 + teams['finals'] * 3 + teams['appearances']
visibility = teams.sort_values('visibility_score', ascending=False)
visibility[['team', 'region', 'titles', 'finals', 'appearances', 'visibility_score']].head(10)
```

**Finding:** Brazil, Germany, Italy, Argentina, and France rank highest because they combine repeated appearances with deep tournament success. This is a stronger portfolio answer than simply saying a team is popular without explaining the measurement.

![World Cup visibility score](assets/charts/world-cup-visibility-score.svg)

## Question 4: Which Regions Dominate World Cup Wins?

```python
region_titles = editions.groupby('winner_region').size().reset_index(name='titles').sort_values('titles', ascending=False)
region_titles
```

**Finding:** Europe leads total titles, but South America remains extremely strong because Brazil, Argentina, and Uruguay have won multiple tournaments.

![World Cup titles by region](assets/charts/world-cup-region-titles.svg)

## Question 5: Do Highest-Paid Players Come From World Cup Powers?

```python
player_context = players.merge(
    teams[['team', 'titles', 'finals', 'appearances']],
    left_on='country',
    right_on='team',
    how='left'
).drop(columns=['team'])

player_context[['rank', 'player', 'country', 'club', 'total_earnings_musd', 'titles', 'finals', 'appearances']]
```

**Finding:** Individual earnings and national-team success are related, but not the same thing. Messi, Mbappe, Benzema, Vinicius, Bellingham, and Yamal come from World Cup-winning countries. Ronaldo, Haaland, Salah, and Mane show that individual market value can be very high even when the player's country has fewer World Cup titles or no titles.

![Highest-paid players with World Cup context](assets/charts/highest-paid-player-context.svg)

## Conclusion

Brazil is the most successful World Cup country by titles, while Germany stands out for finals consistency. A visibility-score approach shows that the most globally visible World Cup teams are not only teams with trophies, but teams that repeatedly qualify and reach late tournament stages.

The player earnings analysis adds a business layer: football pay reflects club economics, sponsorships, global brand value, and league strategy, not only national-team results.
